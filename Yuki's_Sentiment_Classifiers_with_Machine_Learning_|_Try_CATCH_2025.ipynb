## **Sentiment Classifier**

### **Import Libraries**

This process enables us to leverage pre-written functions and classes without needing to write them ourselves.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

### **Load the Dataset**

A **dataset** is a table full of information. Each row is one example.

In this case:
- One column is the **review text**
- The other column is the **label** (positive or negative)

These are our training examples!

```python
# Load the dataset
df = pd.read_csv("https://raw.githubusercontent.com/cancode-teacher/dataset-files/main/amazon_reviews.csv")

# Show the first few rows to see what the data looks like
df.head()

```

### **Let's Build Our Machine Learning Model!**

A Machine Learning **model** is a smart tool that:
1. Looks at patterns in your training data (review + label)
2. Learns rules that connect inputs to outputs
3. Uses those rules to make predictions on new data!

Think of it like this:
- You show your friend 100 reviews labeled as "positive" or "negative"
- After reading enough, they start to guess the label just by looking at the words
- That’s what your model does!

Now let’s help it learn!

### **Why Split?**

We split our data so we can train the model on one part and test it on another.  
That way, we can see if it actually **learned** or if it's just memorizing.

```python
# Split the data into two parts:
# - 70% to train the model
# - 30% to test how well it learned

X_train, X_test, y_train, y_test = train_test_split(
    df["reviewText"],
    df["Positive"],
    test_size = 0.3,
    random_state = 42
)
```

### **What is Vectorization?**

Machines cannot understand words or strings. When dealing with text data, we need to represent it in numerical form to be understood by the machine.

"Vectorizing" text means turning words into numbers so the model can understand them.

CountVectorizer (a class in the scikit-learn Python library) creates a big table where each column is a word and each row counts how many times each word appears in a review.

For example, if your review was `this is so good, and bad, but mostly bad`:


*   CountVectorizer creates a table with the following columns: this, is, so, good, and, bad, but, mostly
*   For each word column, there is an integer count representing the frequency of the word in a review
  * Note: the word `bad` has a frequency count of 2 and there is only one `bad` column
* With a lot of examples, our model can recognize patterns between the frequency of certain words and labels!
   * For example, our model can learn that multiple mentions of the word "bad" is commonly associated with the Negative label

More information on CountVectorizer can be found here:

*   [Geeks for Geeks](https://www.geeksforgeeks.org/nlp/using-countvectorizer-to-extracting-features-from-text/)
*   [Towards Data Science](https://towardsdatascience.com/basics-of-countvectorizer-e26677900f9c/)



```python
# ML models can't understand text, so we need to turn words into numbers

# Count how often each word appears
vectorizer = CountVectorizer()

# Learn and transform the training data
X_train_vec = vectorizer.fit_transform(X_train)

# Only transform the test data
X_test_vec = vectorizer.transform(X_test)
```

### **What is Training?**

Training is when the model looks at the inputs (reviews) and outputs (labels), and figures out the patterns that connect them.

MultinomialNB (a class in the scikit-learn Python library) is a classifier that's designed for data where features are represented by counts. Since we are counting how many times each word is found in a review, this classifier is perfect! The classifier will learn the probability of each word occurring within each class (Positive/Negative).

More information on MultinomialNB can be found here:

*   [Geeks for Geeks](https://www.geeksforgeeks.org/machine-learning/multinomial-naive-bayes/)

```python
# Create a Naive Bayes classifier (a simple but powerful model for text)
model = MultinomialNB()

# Train the model using our training data
model.fit(X_train_vec, y_train)
```

### **Make Predictions!**

```python
# Label mapping dictionary
label_map = {0:"negative", 1:"positive"}

# Use the model to predict the labels of the test data
y_pred = model.predict(X_test_vec)

# Print out some predictions
print("Example predictions:")
for i in range(5):
  print(f"Review: {X_test.iloc[i]}")
  print(f"Actual: {label_map[y_test.iloc[i]]}, Predicted: {label_map[y_pred[i]]}")
  print("------")
```

### **What is Accuracy?**

Accuracy is how many times the model got the answer right.  
If it got 80 out of 100 reviews correct, the accuracy is 80%.

### **Try It Yourself!**

```python
# Try writing your own review and seeing what the model thinks!
our_review = input("Type a review: ")

# Vectorize the input (wrap it in a list)
your_review_vec = vectorizer.transform([our_review])

# Predict sentiment
prediction = model.predict(your_review_vec)

# Print the result
print("Your Review:", our_review)
print("Predicted Sentiment:", label_map[prediction[0]])
```
